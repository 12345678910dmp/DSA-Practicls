```python
# Problem statement 1.1: Array Rotation

n = int(input("Enter number of items: "))
arr = list(map(int, input("Enter items: ").split()))
h = int(input("Enter number of hours: "))

# Effective rotations
h = h % n

# Rotate left by h positions
arr = arr[h:] + arr[:h]

print("Final display order:", arr)
```

```python
# Problem statement 1.2: Frequency Analysis

n = int(input("Enter number of book records: "))
arr = list(map(int, input("Enter book IDs: ").split()))

frequency = {}

# Count frequency of each book ID
for book_id in arr:
    if book_id in frequency:
        frequency[book_id] += 1
    else:
        frequency[book_id] = 1

print("Books borrowed more than once:")

for book_id in frequency:
    if frequency[book_id] > 1:
        print(book_id, end=" ")
```

```python
# Problem statement 1.3: Find Longest Word

sentence = input("Enter a sentence: ")

words = sentence.split()

longest_word = ""

for word in words:
    if len(word) > len(longest_word):
        longest_word = word

print("Longest word:", longest_word)
print("Length:", len(longest_word))
```
