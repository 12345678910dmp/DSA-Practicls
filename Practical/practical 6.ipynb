```python
# Problem statement 6.1: Stack Using Array

class ArrayStack:
    def __init__(self, capacity):
        self.stack = [None] * capacity
        self.capacity = capacity
        self.top = -1

    # Push operation
    def push(self, data):
        # Check if stack is full
        if self.top == self.capacity - 1:
            print("Error: Stack Overflow - Stack is full")
            return

        self.top += 1
        self.stack[self.top] = data

        print(data, "placed on stack")

    # Pop operation
    def pop(self):
        # Check if stack is empty
        if self.top == -1:
            print("Error: Stack Underflow - Stack is empty")
            return

        data = self.stack[self.top]
        self.stack[self.top] = None
        self.top -= 1

        print(data, "taken from stack")

    # Display top element
    def peek(self):
        if self.top == -1:
            print("Stack is empty")
        else:
            print("Top tray:", self.stack[self.top])

    # Display stack
    def display(self):
        if self.top == -1:
            print("Stack is empty")
            return

        print("Stack:", end=" ")

        for i in range(self.top, -1, -1):
            print(self.stack[i], end=" ")

        print()


# Main Program

capacity = int(input("Enter stack capacity: "))

stack = ArrayStack(capacity)

stack.push(101)
stack.peek()

stack.push(102)
stack.peek()

stack.push(103)
stack.peek()

stack.display()

stack.pop()
stack.peek()

stack.pop()
stack.peek()

stack.display()
```

```python
# Problem statement 6.2: Stack Using Linked List

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedStack:
    def __init__(self):
        self.top = None

    # Push operation
    def push(self, data):
        new_node = Node(data)

        new_node.next = self.top
        self.top = new_node

        print(data, "visited")

    # Pop operation
    def pop(self):
        if self.top is None:
            print("No previous page - History is empty")
            return

        data = self.top.data
        self.top = self.top.next

        print("Back to:", self.top.data if self.top else "No page")

    # Display current page
    def peek(self):
        if self.top is None:
            print("No page is currently open")
        else:
            print("Current page:", self.top.data)

    # Display history
    def display(self):
        if self.top is None:
            print("History is empty")
            return

        current = self.top

        print("History:", end=" ")

        while current is not None:
            print(current.data, end=" -> ")
            current = current.next

        print("None")


# Main Program

browser = LinkedStack()

browser.push("Google")
browser.peek()

browser.push("YouTube")
browser.peek()

browser.push("Wikipedia")
browser.peek()

browser.display()

browser.pop()
browser.peek()

browser.pop()
browser.peek()

browser.display()

browser.pop()
browser.pop()
```

```python
# Problem statement 6.3: Infix to Postfix

def precedence(operator):
    if operator == '+' or operator == '-':
        return 1

    if operator == '*' or operator == '/':
        return 2

    if operator == '^':
        return 3

    return 0


def infix_to_postfix(expression):

    stack = []
    postfix = []

    for character in expression:

        # Ignore spaces
        if character == ' ':
            continue

        # Operand
        if character.isalnum():
            postfix.append(character)

        # Opening bracket
        elif character == '(':
            stack.append(character)

        # Closing bracket
        elif character == ')':

            while stack and stack[-1] != '(':
                postfix.append(stack.pop())

            # Remove '('
            if stack:
                stack.pop()

        # Operator
        else:

            while (stack and
                   stack[-1] != '(' and
                   precedence(stack[-1]) >= precedence(character)):

                postfix.append(stack.pop())

            stack.append(character)

    # Pop remaining operators
    while stack:
        postfix.append(stack.pop())

    return " ".join(postfix)


# Main Program

expression = input("Enter infix expression: ")

result = infix_to_postfix(expression)

print("Postfix expression:", result)
```
