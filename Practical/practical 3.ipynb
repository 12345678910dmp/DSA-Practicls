```python
# problem statement 3.1: Sorting Algorithms

# Bubble Sort
def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        swapped = False

        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                # Swap adjacent elements
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True

        # If no swapping happened, array is already sorted
        if not swapped:
            break

    return arr


# Selection Sort
def selection_sort(arr):
    n = len(arr)

    for i in range(n):
        min_index = i

        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j

        # Place smallest element at current position
        arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr


# Insertion Sort

def insertion_sort(arr):
    n = len(arr)

    for i in range(1, n):
        key = arr[i]
        j = i - 1

        # Move larger elements one position ahead
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        # Insert key at correct position
        arr[j + 1] = key

    return arr


# Main Program
 
marks = list(map(int, input("Enter student marks: ").split()))

# Make separate copies
bubble = marks.copy()
selection = marks.copy()
insertion = marks.copy()

# Apply sorting
bubble_sort(bubble)
selection_sort(selection)
insertion_sort(insertion)

# Display results
print("\nOriginal marks:", marks)

print("Bubble Sort:", bubble)
print("Selection Sort:", selection)
print("Insertion Sort:", insertion)
```

```python
# Problem statement 3.2: Sort 0s, 1s and 2s
# Dutch National Flag Algorithm

def sort_colors(arr):
    low = 0
    mid = 0
    high = len(arr) - 1

    while mid <= high:

        if arr[mid] == 0:
            # Put 0 at the low position
            arr[low], arr[mid] = arr[mid], arr[low]

            low += 1
            mid += 1

        elif arr[mid] == 1:
            # 1 is already in the correct middle region
            mid += 1

        else:
            # Put 2 at the high position
            arr[mid], arr[high] = arr[high], arr[mid]

            high -= 1

    return arr


# Main Program

arr = list(map(int, input("Enter color codes (0, 1, 2): ").split()))

sort_colors(arr)

print("Sorted colors:", arr)
```
