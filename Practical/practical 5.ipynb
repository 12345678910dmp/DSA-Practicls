```python
# Problem statement 5.1: Doubly Linked List
# Music Playlist

class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


class DoublyLinkedList:
    def __init__(self):
        self.head = None

    # Insert at beginning
    def insert_beginning(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        new_node.next = self.head
        self.head.prev = new_node
        self.head = new_node

    # Insert at end
    def insert_end(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head

        while current.next is not None:
            current = current.next

        current.next = new_node
        new_node.prev = current

    # Insert after a specific song
    def insert_after(self, song, data):
        current = self.head

        while current is not None:

            if current.data == song:
                new_node = Node(data)

                # Connect new node to next node
                new_node.next = current.next
                new_node.prev = current

                # If current is not the last node
                if current.next is not None:
                    current.next.prev = new_node

                # Connect current to new node
                current.next = new_node

                print(data, "inserted after", song)
                return

            current = current.next

        print(song, "not found")

    # Delete first song
    def delete_first(self):
        if self.head is None:
            print("Playlist is empty")
            return

        removed = self.head.data

        self.head = self.head.next

        if self.head is not None:
            self.head.prev = None

        print(removed, "removed")

    # Count songs
    def count(self):
        count = 0
        current = self.head

        while current is not None:
            count += 1
            current = current.next

        return count

    # Display playlist
    def display(self):
        current = self.head

        print("Playlist:", end=" ")

        while current is not None:
            print(current.data, end=" <-> ")
            current = current.next

        print("None")


# Main Program

playlist = DoublyLinkedList()

print("Insert at beginning:")
playlist.insert_beginning("Song A")
playlist.display()

playlist.insert_beginning("Song B")
playlist.display()

print("\nInsert at end:")
playlist.insert_end("Song C")
playlist.display()

playlist.insert_end("Song D")
playlist.display()

print("\nInsert after Song B:")
playlist.insert_after("Song B", "Song X")
playlist.display()

print("\nCount:")
print("Total songs:", playlist.count())

print("\nDelete first song:")
playlist.delete_first()
playlist.display()

print("\nFinal count:")
print("Total songs:", playlist.count())

print("\nTrying to insert after a song that does not exist:")
playlist.insert_after("Song Z", "Song Y")
playlist.display()
```

```python
# Problem statement 5.2.1

# Singly Circular Linked List

class SNode:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyCircularList:
    def __init__(self):
        self.head = None

    # Insert at end
    def insert_end(self, data):
        new_node = SNode(data)

        if self.head is None:
            self.head = new_node
            new_node.next = self.head
            return

        current = self.head

        while current.next != self.head:
            current = current.next

        current.next = new_node
        new_node.next = self.head

    # Insert at beginning
    def insert_beginning(self, data):
        new_node = SNode(data)

        if self.head is None:
            self.head = new_node
            new_node.next = self.head
            return

        current = self.head

        while current.next != self.head:
            current = current.next

        new_node.next = self.head
        current.next = new_node
        self.head = new_node

    # Delete a student by value
    def delete(self, data):

        if self.head is None:
            print("Circle is empty")
            return

        # Only one node
        if self.head.data == data and self.head.next == self.head:
            self.head = None
            print(data, "left the circle")
            return

        # Delete head
        if self.head.data == data:
            current = self.head

            while current.next != self.head:
                current = current.next

            current.next = self.head.next
            self.head = self.head.next

            print(data, "left the circle")
            return

        # Delete middle/end node
        current = self.head

        while current.next != self.head:

            if current.next.data == data:
                current.next = current.next.next
                print(data, "left the circle")
                return

            current = current.next

        print(data, "not found")

    # Display circle
    def display(self):
        if self.head is None:
            print("Circle is empty")
            return

        current = self.head

        print("Circle:", end=" ")

        while True:
            print(current.data, end=" -> ")
            current = current.next

            if current == self.head:
                break

        print("(back to", self.head.data, ")")
```

```python
# Problem statement 5.2.2

# Doubly Circular Linked List

class DNode:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


class DoublyCircularList:
    def __init__(self):
        self.head = None

    # Insert at end
    def insert_end(self, data):
        new_node = DNode(data)

        if self.head is None:
            self.head = new_node

            new_node.next = new_node
            new_node.prev = new_node

            return

        last = self.head.prev

        new_node.next = self.head
        new_node.prev = last

        last.next = new_node
        self.head.prev = new_node

    # Insert at beginning
    def insert_beginning(self, data):
        new_node = DNode(data)

        if self.head is None:
            self.head = new_node

            new_node.next = new_node
            new_node.prev = new_node

            return

        last = self.head.prev

        new_node.next = self.head
        new_node.prev = last

        last.next = new_node
        self.head.prev = new_node

        self.head = new_node

    # Delete a student
    def delete(self, data):

        if self.head is None:
            print("Circle is empty")
            return

        current = self.head

        while True:

            if current.data == data:

                # Only one node
                if current.next == current:
                    self.head = None

                else:
                    current.prev.next = current.next
                    current.next.prev = current.prev

                    # If deleting head
                    if current == self.head:
                        self.head = current.next

                print(data, "left the circle")
                return

            current = current.next

            if current == self.head:
                break

        print(data, "not found")

    # Display circle
    def display(self):
        if self.head is None:
            print("Circle is empty")
            return

        current = self.head

        print("Circle:", end=" ")

        while True:
            print(current.data, end=" <-> ")
            current = current.next

            if current == self.head:
                break

        print("(back to", self.head.data, ")")
```
