```python
# Problem Statement 2.1: Linear Search
# Iterative and Recursive

# Iterative Linear Search
def linear_search_iterative(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1


# Recursive Linear Search
def linear_search_recursive(arr, target, index=0):
    # Base case
    if index >= len(arr):
        return -1

    # Target found
    if arr[index] == target:
        return index

    # Recursive call
    return linear_search_recursive(arr, target, index + 1)


# Input
arr = list(map(int, input("Enter license plates: ").split()))
target = int(input("Enter target plate: "))

# Iterative search
result1 = linear_search_iterative(arr, target)

# Recursive search
result2 = linear_search_recursive(arr, target)

print("Iterative Linear Search Position:", result1)
print("Recursive Linear Search Position:", result2)
```

```python
# Problem Statement 2.2: Binary Search
# Iterative and Recursive

# Iterative Binary Search
def binary_search_iterative(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid

        elif arr[mid] < target:
            low = mid + 1

        else:
            high = mid - 1

    return -1


# Recursive Binary Search
def binary_search_recursive(arr, target, low, high):
    # Base case
    if low > high:
        return -1

    mid = (low + high) // 2

    if arr[mid] == target:
        return mid

    elif arr[mid] < target:
        return binary_search_recursive(arr, target, mid + 1, high)

    else:
        return binary_search_recursive(arr, target, low, mid - 1)


# Input
arr = list(map(int, input("Enter sorted book codes: ").split()))
target = int(input("Enter target book code: "))

# Iterative search
result1 = binary_search_iterative(arr, target)

# Recursive search
result2 = binary_search_recursive(arr, target, 0, len(arr) - 1)

print("Iterative Binary Search Position:", result1)
print("Recursive Binary Search Position:", result2)
```
