```python
# Problem statement 7.1

class CircularQueue:
    def __init__(self, capacity):
        self.capacity = capacity
        self.queue = [None] * capacity
        self.front = 0
        self.rear = -1
        self.count = 0

    # Join the queue
    def enqueue(self, token):
        if self.count == self.capacity:
            print("Error: Queue Overflow - Queue is full")
            return

        self.rear = (self.rear + 1) % self.capacity
        self.queue[self.rear] = token
        self.count += 1

        print(token, "joined the queue")

    # Serve the front person
    def dequeue(self):
        if self.count == 0:
            print("Error: Queue Underflow - Queue is empty")
            return

        token = self.queue[self.front]
        self.queue[self.front] = None

        self.front = (self.front + 1) % self.capacity
        self.count -= 1

        print(token, "was served")

    # Show front token
    def peek(self):
        if self.count == 0:
            print("Queue is empty")
        else:
            print("Front token:", self.queue[self.front])

    # Display queue
    def display(self):
        if self.count == 0:
            print("Queue is empty")
            return

        print("Queue:", end=" ")

        index = self.front

        for i in range(self.count):
            print(self.queue[index], end=" ")
            index = (index + 1) % self.capacity

        print()


# Main program
capacity = int(input("Enter queue capacity: "))

q = CircularQueue(capacity)

q.enqueue(101)
q.enqueue(102)
q.enqueue(103)

q.peek()
q.display()

q.dequeue()

q.peek()
q.display()

q.enqueue(104)
q.enqueue(105)

q.peek()
q.display()

q.enqueue(106)   # Tests full condition
```

```python
# Problem statement 7.2


class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedQueue:
    def __init__(self):
        self.front = None
        self.rear = None

    # Add patient
    def enqueue(self, patient):
        new_node = Node(patient)

        # If queue is empty
        if self.rear is None:
            self.front = new_node
            self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node

        print(patient, "arrived")

    # Attend front patient
    def dequeue(self):
        if self.front is None:
            print("Error: Queue Underflow - No patients waiting")
            return

        patient = self.front.data
        self.front = self.front.next

        # If queue becomes empty
        if self.front is None:
            self.rear = None

        print(patient, "was attended")

    # Show front patient
    def peek(self):
        if self.front is None:
            print("Ward is empty")
        else:
            print("Front patient:", self.front.data)

    # Display all patients
    def display(self):
        if self.front is None:
            print("Ward is empty")
            return

        current = self.front

        print("Waiting patients:", end=" ")

        while current is not None:
            print(current.data, end=" -> ")
            current = current.next

        print("None")


# Main program
q = LinkedQueue()

q.enqueue("Patient A")
q.enqueue("Patient B")
q.enqueue("Patient C")

q.peek()
q.display()

q.dequeue()

q.peek()
q.display()

q.enqueue("Patient D")

q.peek()
q.display()

q.dequeue()
q.dequeue()
q.dequeue()

q.dequeue()     # Tests empty queue
```
